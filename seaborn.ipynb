violin plot use seaborn libray

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
sns.set(style="whitegrid")
x=sns.load_dataset("tips")
x
sns.violinplot(x="tip",y="total_bill",data=x,hue="time",linewidth=2,palette="Dark2",saturation=0.5,color="red")
plt.show()
```

```python
sns.violinplot(x="time",y="tip",data=x,order=["Dinner","Lunch"],inner="quart")
plt.show()
```

```python
sns.violinplot(x="time",y="tip",data=x,hue="sex",order=["Dinner","Lunch"])
plt.show()
```

```python
sns.violinplot(y=x["total_bill"])
plt.show()
```

pair polt in seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd
df = sns.load_dataset("tips")
sns.pairplot(df,hue="sex",hue_order=["Famle","Male"],palette="BuGn",x_vars=["total_bill","tip"],kind="reg")
plt.show() # kind=change in data of ("scatter",kde,hist,reg)\ markers= data in change of simbel(<,>)
```

strip plot in seaborn

```python
import matplotlib.pyplot as plt # jitter= in data me shepes karne ke liye hota hai
import seaborn as sns # edgecolor =  in ues to color dene ke liye hota hai
import pandas as pd # linewidth = data ka line incress karne ke liye in use
df = sns.load_dataset("tips") # palette= color dene ke liye hota hai in ues
sns.stripplot(x="day",y="total_bill",data=df,hue="sex",palette="rocket_r",hue_order=["Famel","Male"],
              linewidth=1,edgecolor="r",jitter=5,size=4,marker=">",alpha=0.5)
plt.show()
```

```python
sns.stripplot(y=df["day"])
plt.show()
```
