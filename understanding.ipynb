```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
tips= sns.load_dataset('tips')
tips.head(20)
```

```python
# titanic= pd.read_csv("C:\\Users\\ABC\\AppData\\Local\\Temp\\Rar$DIa23980.8508\\insurance.csv")
# titanic
```

scatterplot (Numerical-numerical)

```python
sns.scatterplot(x='total_bill',y='tip',data=tips,hue='sex',style=tips['smoker'])
```

```python
tips.head()
```

Bar plot (Numerical-categorical)

```python
sns.barplot(x='total_bill',y='tip',data=tips,hue=tips['sex'])
```

Box plot (Nunmerical-categorical)

```python
sns.boxplot(x='sex',y='tip',data=tips)
```

Distplot (Numerical-categorical)

```python
sns.distplot(tips['total_bill'])
```

heatmap (categorical-categorical)

```python
tips.head()
```

```python
sns.heatmap(pd.crosstab(tips['total_bill'], tips['tip']))
```

```python
tips.groupby('total_bill').mean(['tip'])
```

categorical-categorical

```python
sns.clustermap(pd.crosstab(tips['total_bill'],tips['tip']))
```

pairplot

```python
sns.pairplot(tips,hue='sex')
```

lineplot (Numerical-numerical)

```python
tips.head()
```

```python
sns.lineplot(x='total_bill',y='tip',data=tips)
```

```python
sns.heatmap(tips.pivot_table(values='tip',index='total_bill',columns='day'))
```
