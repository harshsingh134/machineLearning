```python
import matplotlib.pyplot as plt

# Example loss values (epoch-wise)
loss = [0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2]
less = [0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1]

# Epochs
epochs = range(1, len(loss) + 1)

# Plotting the loss
plt.plot(epochs, loss, marker='o', linestyle='-', color='red', label='Training Loss')
plt.plot(epochs, less, marker='o', linestyle='-', color='red')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Training Loss over Epochs')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sample data
data = [10, 12, 15, 18, 20, 22, 25, 28, 30, 100]

# Histogram (Distribution)
plt.figure(figsize=(10,4))

plt.subplot(1, 2, 1)
plt.hist(data, bins=5)
plt.title("Data Distribution (Histogram)")
plt.xlabel("Value")
plt.ylabel("Frequency")

# Boxplot (Outlier Detection)
plt.subplot(1, 2, 2)
plt.boxplot(data)
plt.title("Outlier Detection (Boxplot)")

plt.tight_layout()
plt.show()
```
