```python
unique_emotions = df['emoting'].unique()
emotion_number = {}
i = 0
for emo in unique_emotions:
    emotion_number[emo] = i
    i += 1

    df['emoting'] = df['emotion'].map(emotion_number)
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
import string
def remove_punctuation(txt):
    return txt.translate(str.maketrans('', '', string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punctuation)
```

```python
def remove_numders(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new += i
            return new
        df['text'] = df['text'].apply(remove_numders)  
```
